```python
1️⃣ np.array()
```

```python
#Purpose: Create a NumPy array.
import numpy as np

arr = np.array([1, 2, 3, 4])
print(arr)
```

```python
2️⃣ np.arrange()
```

```python
#Purpose: Create evenly spaced values within a range (like range() in Python but returns an array).
    
arr = np.arange(0, 10, 2)
print(arr)
```

```python
3️⃣ np.linspace()
```

```python
#Purpose: Create an array with evenly spaced numbers between two limits.

arr = np.linspace(0, 1, 5)
print(arr)
```

```python
4️⃣ np.zeros() / np.ones()
```

```python
#Purpose: Create arrays filled with zeros or ones.

zeros = np.zeros((2, 3))
ones = np.ones((2, 3))

print(zeros)
print(ones)
```

```python
5️⃣ np.eye()
```

```python
#Purpose: Create an identity matrix.

arr = np.eye(3)
print(arr)
```

```python
6️⃣ np.random.rand() / np.random.randn()
```

```python
#Purpose: Generate random arrays.
#rand() → uniform distribution
#randn() → normal distribution


print(np.random.rand(2, 3))
print(np.random.randn(2, 3))
```

```python
7️⃣ arr.reshape()
```

```python
#Purpose: Change the shape of an array.

arr = np.arange(6)
reshaped = arr.reshape(2, 3)
print(reshaped)
```

```python
8️⃣ arr.ndim, arr.shape, arr.size
```

```python
#Purpose: Get number of dimensions, shape, and total elements.

arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.ndim)   # Number of dimensions
print(arr.shape)  # Shape (rows, columns)
print(arr.size)   # Total elements
```

```python
9️⃣ np.concatenate()
```

```python
#Purpose: Join two or more arrays.

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.concatenate((a, b)))
```

```python
🔟 np.vstack() / np.hstack()
```

```python
#Purpose: Stack arrays vertically or horizontally.

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.vstack((a, b)))
print(np.hstack((a, b)))
```

```python
1️⃣1️⃣ arr.T
```

```python
#Purpose: Transpose of an array (rows → columns).
    
arr = np.array([[1, 2], [3, 4]])
print(arr.T)
```

```python
1️⃣2️⃣ np.sum(), np.mean(), np.std()
```

```python
#Purpose: Basic statistics.

arr = np.array([1, 2, 3, 4])
print(np.sum(arr))
print(np.mean(arr))
print(np.std(arr))
```

```python
1️⃣3️⃣ np.min(), np.max(), np.argmin(), np.argmax()
```

```python
#Purpose: Find min/max values and their indices.

arr = np.array([4, 8, 1, 6])
print(np.min(arr))
print(np.max(arr))
print(np.argmin(arr))
print(np.argmax(arr))
```

```python
1️⃣4️⃣ np.sqrt(), np.exp(), np.log()
```

```python
#Purpose: Element-wise math operations.

arr = np.array([1, 4, 9])
print(np.sqrt(arr))
print(np.exp(arr))
print(np.log(arr))
```

```python
1️⃣5️⃣ np.dot() / np.matmul()
```

```python
2️⃣0️⃣ np.delete()
```

```python
2️⃣1️⃣ np.reshape() vs np.ravel() vs np.flatten()
```

```python
#Purpose:
#reshape() → Changes shape
#Travel() / flatten() → Converts to 1D

arr = np.array([[1, 2], [3, 4]])
print(arr.ravel())
print(arr.flatten())
```

```python
2️⃣2️⃣ np.save() / np.load()
```

```python
#Purpose: Save and load NumPy arrays.

arr = np.array([1, 2, 3])
np.save('my_array.npy', arr)
loaded = np.load('my_array.npy')
print(loaded)
```

```python
2️⃣3️⃣ np.all() / np.any()
```

```python
#Purpose: Logical checks.

arr = np.array([1, 0, 3])
print(np.all(arr))  # True if all non-zero
print(np.any(arr))  # True if any non-zero
```
